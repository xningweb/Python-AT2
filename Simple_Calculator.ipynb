### Simple Calculator

The Simple Calculator is a program that allows users to perform basic mathematical calculations.
The program displays a menu with five options, including addition, subtraction, multiplication, division and exit.
The user selects an operation and enters two numbers. The program calculates and displays the result.
The program uses a loop to allow users to perform multiple calculations until they choose to exit.
It also handles invalid menu selections, non-numeric inputs and division by zero.
A list is used to store the calculation history.


### Algorithm for Simple Calculator

1. Start the program.
2. Create an empty list to store the calculation history.
3. Display the calculator menu with five options.
4. Ask the user to select an option.
5. If the user selects option 5, display an exit message and end the program.
6. If the user selects an invalid option, display an error message and return to the menu.
7. If the user selects an operation from 1 to 4, ask for two numbers.
8. Check whether both inputs are valid numbers.
9. If the inputs are invalid, display an error message and return to the menu.
10. Perform the selected mathematical operation using a function.
11. If division is selected, check that the second number is not zero.
12. Display the calculation result formatted to two decimal places.
13. Store the successful calculation in a list.
14. Return to the main menu.
15. Repeat until the user selects option 5.
16. End the program.


### Pseudocode for Simple Calculator

Create empty calculation history list

While True

	Display “Simple Calculator”
	Display “1. Addition”
	Display “2. Subtraction”
	Display “3. Multiplication”
	Display “4. Division”
	Display “5. Exit”

	Input choice

	If choice = 5 Then
		Display “Goodbye!”
		Break
	Else if choice is between 1 and 4 then
		Input first_number
		Input second_number

		If either input is not a valid number then
			Display “Invalid number”
			Continue
		End if

		If choice = 4 and second_number = 0 then
			Display “Cannot divide by zero”
			Continue
		End if

		Call calculate function
		Display result to two decimal places
		Add calculation to calculation_history
	Else
		Display “Invalid choice”
	End if

End while


```python
# Task 3: Simple Calculator
# This program performs basic mathematical operations.

import math

# Function to perform a calculation
def calculate(num1, num2, choice):
  if choice == "1":
    return num1 + num2
  elif choice == "2":
    return num1 - num2
  elif choice == "3":
    return num1 * num2
  elif choice == "4":
    return num1 / num2

# Create an empty list to store calculation history
calculation_history = []

# Main program loop
while True:
  print("\n==== SIMPLE CALCULATOR ====")
  print("1. Addition")
  print("2. Subtraction")
  print("3. Multiplication")
  print("4. Division")
  print("5. Exit")

  # Ask the user to choose an option
  choice = input("Enter your choice (1-5): ")

  # Exit the program
  if choice == "5":
    print("Thank you for using Simple Calculator!")
    break

  # Check if the menu choice is valid
  if choice not in ["1", "2", "3", "4"]:
    print("Invalid choice. Please select 1-5.")
    continue

  # Get two numbers from the user
  try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
  except ValueError:
    print("Invalid input. Please enter numbers only.")
    continue

  # Check for non-finite values
  if not math.isfinite(num1) or not math.isfinite(num2):
    print("Invalid input. Please enter finite numbers.")
    continue

  # Prevent division by zero
  if choice == "4" and num2 == 0:
    print("Error: Cannot divide by zero.")
    continue

  # Perform the calculation using the function
  try:
    result = calculate(num1, num2, choice)
  except OverflowError:
    print("Error: Calculation result is too large.")
    continue

  # Check whether the result is valid
  if not math.isfinite(result):
    print("Error: Calculation result is too large.")
    continue

  # Display the result with two decimal places
  print(f"Result: {result:.2f}")

  # Save the calculation result in a list
  calculation_history.append(result)

# Display the calculation history after exiting
print("\n==== CALCULATION HISTORY ==== ")

if len(calculation_history) == 0:
  print("No calculations were performed.")
else:
  for index, result in enumerate(calculation_history, start=1):
    print(f"Calculation {index}:{result:.2f}")


print("Program ended.")
```
