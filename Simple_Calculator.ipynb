```python
# Task 3: Simple Calculator
# This program performs basic mathematical operations.

import math

# Function to perform a calculation
def calculate(num1, num2, choice):
  if choice == "1":
    return num1 + num2
  elif choice == "2":
    return num1 - num2
  elif choice == "3":
    return num1 * num2
  elif choice == "4":
    return num1 / num2

# Create an empty list to store calculation history
calculation_history = []

# Main program loop
while True:
  print("\n==== SIMPLE CALCULATOR ====")
  print("1. Addition")
  print("2. Subtraction")
  print("3. Multiplication")
  print("4. Division")
  print("5. Exit")

  # Ask the user to choose an option
  choice = input("Enter your choice (1-5): ")

  # Exit the program
  if choice == "5":
    print("Thank you for using Simple Calculator!")
    break

  # Check if the menu choice is valid
  if choice not in ["1", "2", "3", "4"]:
    print("Invalid choice. Please select 1-5.")
    continue

  # Get two numbers from the user
  try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
  except ValueError:
    print("Invalid input. Please enter numbers only.")
    continue

  # Check for non-finite values
  if not math.isfinite(num1) or not math.isfinite(num2):
    print("Invalid input. Please enter finite numbers.")
    continue

  # Prevent division by zero
  if choice == "4" and num2 == 0:
    print("Error: Cannot divide by zero.")
    continue

  # Perform the calculation using the function
  try:
    result = calculate(num1, num2, choice)
  except OverflowError:
    print("Error: Calculation result is too large.")
    continue

  # Check whether the result is valid
  if not math.isfinite(result):
    print("Error: Calculation result is too large.")
    continue

  # Display the result with two decimal places
  print(f"Result: {result:.2f}")

  # Save the calculation result in a list
  calculation_history.append(result)

# Display the calculation history after exiting
print("\n==== CALCULATION HISTORY ==== ")

if len(calculation_history) == 0:
  print("No calculations were performed.")
else:
  for index, result in enumerate(calculation_history, start=1):
    print(f"Calculation {index}:{result:.2f}")


print("Program ended.")
```
