```python
# Task 1a: Create a text file containing names

with open("names.txt","w") as file:
    file.write("Luke Skywalker\n")
    file.write("Leia Organa\n")
    file.write("Han Solo\n")
    file.write("Obi-Wan Kenobi\n")

print("names.txt has been created sucessfully.")
```

```python
# Read and display each line from the text file

with open("names.txt", "r") as file:
  for line in file:
    print(line.strip())
```

```python
#Task 1b: Write numerical data to a text file

# Create a list containing five numbers
numbers = [10, 20, 30, 40, 50]

# Open a text file in write mode
with open("numbers.txt", "w")as file:
  for number in numbers:
    file.write(str(number) + "\n")

# Display a confirmation message
print("Numbers have been weitten to numbers.txt successfully.")
```
